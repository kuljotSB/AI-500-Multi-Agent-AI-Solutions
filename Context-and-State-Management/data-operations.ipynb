### Setting up the Environment Variables

```python
import os
from dotenv import load_dotenv

load_dotenv()
redis_hostname = os.getenv("REDIS_HOSTNAME")
redis_password = os.getenv("REDIS_PASSWORD")
```

### Setting up the Redis Client

```python
import redis

r = redis.Redis(
    host = redis_hostname,
    port = 10000,
    ssl = True,
    decode_responses = True,
    password = redis_password
)
```

### Perform Basic Set and Get Operations

```python
# Store a value
result = r.set('user:1001:name', 'Kuljot Bakshi')
print(f"SET operation successful: {result}")
```

```python
# Retrieve the value
name = r.get('user:1001:name')
print(f"Retrieved name: {name}")
```

```python
# Check if a key exists
if r.exists('user:1001:name'):
    print("Key exists")
```

```python
# Set a key with 60-second expiration
r.set('user:1001:email', 'kuljotbakshi@example.com', ex=60)

# Alternative: Set value with expiration in milliseconds
r.set('user:1001:age', '21', px=5000)  # Expires in 5000ms

# Set expiration to existing keys
r.expire('user:1001:name', 3600)
```

```python
# Delete the value
keys = ['user:1001:name', 'user:1001:email', 'user:1001:age']

for key in keys:
    result = r.delete(key)
    print(f"Keys deleted: {result}")
```

### Store multiple field-value pairs in a Hash

```python
# Store multiple fields in a hash
r.hset('user:1001', mapping={
    'name': 'Alice Smith',
    'email': 'alice@example.com',
    'age': '21'
})

r.hset('user:1002', mapping = {
    'name': 'Hannah Baker',
    'email': 'hannahbaker@example.com',
    'age': '24'
})

# Retrieve a single field
name = r.hget('user:1001', 'name')
print(f"Name: {name}")  # Output: Alice Smith

# Retrieve all fields
user_data = r.hgetall('user:1001')
print(f"User data: {user_data}")  # Returns dict with all fields
```

```python
# Using pipeline for multiple hash operations (efficient)
pipe = r.pipeline(transaction = False)
pipe.hgetall('user:1001')
pipe.hgetall('user:1002')
results = pipe.execute()
print("First pipeline execution results:")
print(results)
print("==================================")
# Returns list of dicts, one for each hash
```

```python
# Or retrieve specific fields from multiple hashes
pipe = r.pipeline(transaction = False)
pipe.hget('user:1001', 'name')
pipe.hget('user:1002', 'name')
names = pipe.execute()
print("Second pipeline execution results:")
print(names)
print("==================================")
# Returns list of names: ['Alice Smith', 'Bob Jones', 'Carol White']
```
